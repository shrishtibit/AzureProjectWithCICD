### Silver Notebook

### Reading nocs data

```python
df = spark.read.format("csv")\
         .option("header", True)\
         .option("inferSchema", True)\
         .load("abfss://bronze@olympicsprojshrishti.dfs.core.windows.net/nocs")
```

```python
display(df)
```

**Dropping the columns** 

```python
df = df.drop('country')
```

```python
df.display()
```

**Applying transformation on tag column**

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df = df.withColumn('tag', split(col('tag'), '-')[0])
df.display()
```

```python
# Writing the data into silver layer
df.write.format("delta")\
        .mode("append")\
        .option("path","abfss://silver@olympicsprojshrishti.dfs.core.windows.net/nocs")\
        .save()
```

```python
# Creating managed table in unity metastore
df.write.format("delta")\
        .mode("append")\
        .saveAsTable("olympics.silver.nocs_man")
```

```python
# Creating external table in our datalake
# External table will write the data to your datalake but store the table definitions and metadata into metastore
df.write.format("delta")\
        .mode("append")\
        .option("path","abfss://silver@olympicsprojshrishti.dfs.core.windows.net/nocs")\
        .saveAsTable("olympics.silver.nocs")
```
